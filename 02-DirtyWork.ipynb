```python
import random
random_list = [random.random() for _ in range(5_000_000)]
print("Random list length:", len(random_list))
```

```python
def mean_custom(values):
    accumulate = 0 
    for val in values:
        accumulate += val

    return accumulate/len(values) # esto calcula el promedio

def st_dev(values):
    st = 0
    mean_val = mean_custom(values)
    for val in values:
        st += (val - mean_val)**2

    st = st / len(values)

    return st**0.5

    
    
```

```python
print(st_dev(random_list))
```

```python
import numpy as np

np_array = np.array(random_list)

print(f'Std:{np_array.std()}')
```

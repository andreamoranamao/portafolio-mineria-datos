```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/andreamoranamao/portafolio-mineria-datos/refs/heads/main/googleplaystore.csv"

df = pd.read_csv(url)

df.head()
```

### Reto 1:

```python
# convertiendo los valores
def convertir_size(size):
    if size == "Varies with device":
        return np.nan
    elif size.endswith("M"):
        return float(size[:-1])
    elif size.endswith("k"):
        return float(size[:-1]) / 1024
```

```python
df["Size"] = df["Size"].apply(convertir_size)
```

```python
#comprobando tipo de dato
df["Size"].dtype
```

```python
# calculando el promedio del tamano de las apps 

promedio_size = df["Size"].mean()

print("peso promedio:", promedio_size, "MB")
```

### Reto 2:

```python
#conveirte la columna de texto a datetime
#error= convierte cualquier valor qie no se pueda interpretar
df["Last Updated"] = pd.to_datetime(df["Last Updated"], errors="coerce")

#obteneindo el año 
df["Year_Updated"] = df["Last Updated"].dt.year

#contamos las aplicaciones de cada año
df["Year_Updated"].value_counts()
```

```python
ano_actualizado = df["Year_Updated"].value_counts().idxmax()

print("año mas aplicaciones actualizadas:", ano_actualizado)
```

### Reto 3:

```python
cantidad_nan = df["Size"].isna().sum()

print("cant registros vacios:", cantidad_nan)
```

```python
#eliminamos las filas que contienen valor nulo 
df = df.dropna(subset=["Size"])
```

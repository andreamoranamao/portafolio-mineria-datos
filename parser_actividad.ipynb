### Pregunta B: 
La columna Installs tiene texto como "1,000,000+". Un algoritmo no puede procesar eso.
#### a: 
Escriban una instrucción encadenando el método .str.replace() para eliminar el símbolo + y otro para eliminar la coma ,. (Ojo: utilicen regex=False o escapen el símbolo más \+).
#### b: 
Al final de la cadena, utilicen .astype(int) (o float si tienen NaNs) para convertir la columna.
#### c: 
Comprueben el éxito ejecutando df['Installs'].mean(). ¿Cuál es el promedio real de descargas?

```python
# quitando el free para que no de error
df = df[df['Installs'] != 'Free']

# limpiando los datos
df['Installs'] = df['Installs'].str.replace('+', '', regex=False).str.replace(',', '', regex=False).astype(int)

# promedio
df['Installs'].mean()
```

### Pregunta C: 
La columna Price tiene valores como 4.99 y 0. Apliquen la misma lógica arquitectónica que en la pregunta anterior eliminen el símbolo de dólar $ y conviertan la columna usando .astype(float).

```python
# eliminando simbolo $ y convirtiendo a float
df['Price'] = df['Price'].str.replace('$', '', regex=False).astype(float)
```

### 1. espacio vectorial de las imagenes
Una imagen RGB de 256 × 256 píxeles tiene tres canales de color rojo, verde y azul. Matematicamente se representa como un tensor de 256 × 256 × 3. Por lo tanto el arreglo tiene p = 3 dimensiones. Si se convierte en un solo vector, seria


256 x 256 x 3 = 196,608 valores o dimensiones.


### 2. espacio Vectorial del Texto

Las palabras no pueden compararse directamente usando distancias matematicas, por eso se convierten en numeros. Con TF-IDF cada documento se representa como un vector donde cada dimension corresponde a una palabra y su valor indica que tan importante es. Con word embeddings, cada palabra se convierte en un vector de numeros reales. Las palabras con significados o contextos similares tienden a quedar mas cerca dentro del espacio vectorial.

### 3. maldicion de la dimensionalidad

Cuando el número de dimensiones p aumenta a cientos, miles o millones, los datos se vuelven mas dispersos. Esto hace que las distancias entre los usuarios de la matriz X sean menos utiles para identificar cuales son realmente similares. los algoritmos necesitan muchos mas datos para trabajar correctamente. lo que viene siendo la maldición de la dimensionalidad.

Basicamente los datos complejos, como imagenes y texto, pueden convertirse en vectores para que los algoritmos puedan procesarlos. Sin embargo, mientras mas dimensiones tenemos, más difícil se vuelve trabajar con las distancias y encontrar patrones.

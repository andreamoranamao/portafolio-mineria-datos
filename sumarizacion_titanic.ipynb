```python
import pandas as pd

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

## 1. Asimetria (Skewness)

La asimetria sirve para saber hacia que lado esta mas concentrada una distribucion. Si el resultado es positivo significa que la cola de la distribucion se extiende hacia la derecha, y si es negativo se extiende hacia la izquierda.

```python
age_skewness = df["Age"].skew()

print("Asimetria de Age:", age_skewness)
```

### Interpretacion

El resultado de la asimetria de Age es positivo, aproximadamente 0.39. Esto quiere decir que la distribucion tiene una ligera inclinacion hacia la derecha. La mayoria de los pasajeros se encuentran en edades jovenes y adultas, mientras que hay menos pasajeros de edades mas grandes. Por eso la cola de la distribucion se extiende hacia las edades mayores.

## 2. Curtosis (Kurtosis)

La curtosis sirve para ver que tan pesadas son las colas de una distribucion y si existen valores extremos. Un resultado positivo puede indicar que hay mas valores extremos que en una distribucion normal.

```python
fare_kurtosis = df["Fare"].kurt()

print("Curtosis de Fare:", fare_kurtosis)
```

### Interpretacion

El resultado de la curtosis de Fare es positivo y es aproximadamente 34. Esto indica que la distribucion tiene colas pesadas y que existen algunos valores muy alejados del resto. Esto si confirma la existencia de outliers que se pueden encontrar en las tarifas, ya que algunos pasajeros pagaron mucho mas que la mayoria.

## 3. Tabla Cruzada (Cross-Tabulation)

La funcion pd.crosstab() sirve para comparar dos variables y mostrar sus resultados en forma de tabla. En este caso voy a comparar la clase del pasajero con si sobrevivio o no.

```python
tabla_cruzada = pd.crosstab(
    df["Pclass"],
    df["Survived"],
    rownames=["Clase del pasajero"],
    colnames=["Sobrevivio"]
)

tabla_cruzada
```

### Interpretacion

Al cruzar Pclass con Survived se puede ver rapidamente que la clase del pasajero tuvo relacion con la supervivencia. En primera clase sobrevivieron 136 pasajeros y 80 no sobrevivieron. En segunda clase sobrevivieron 87 y 97 no sobrevivieron. En tercera clase solo sobrevivieron 119 pasajeros, mientras que 372 no sobrevivieron.

El patron que se puede observar es que los pasajeros de primera clase tuvieron mejores resultados de supervivencia, mientras que en tercera clase hubo una cantidad mucho mayor de pasajeros que no sobrevivieron.

## Conclusion

Con estas tres tecnicas pude resumir diferentes partes del dataset del Titanic. La asimetria de Age mostro que la distribucion de edades esta ligeramente inclinada hacia las edades mayores. La curtosis de Fare mostro que existen valores extremos en las tarifas. Por ultimo, la tabla cruzada permitio ver rapidamente que la clase del pasajero estaba relacionada con la supervivencia, especialmente al comparar primera y tercera clase.
